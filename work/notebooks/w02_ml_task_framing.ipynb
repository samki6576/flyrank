## 1. My lane as an ML task

In the content-refresh lane, I would frame the problem as a ranking task. Given a set of candidate content items that could be refreshed, the model should rank them by expected uplift in future engagement so the team can prioritize the most promising updates.

This is a ranking problem rather than a simple classification because the action is not just "refresh or not"; it is "which refreshes should we do first?"

## 2. Target or proxy

The target I would use is an observed outcome: the change in post-refresh performance relative to the pre-refresh baseline, such as click-through rate lift over the next few days. If that outcome is not available for every historical case, a practical proxy is whether the refreshed content beat its historical baseline on engagement after publication.

## 3. Success metric

I would use precision@20 as the main metric. A good model should put the most promising refresh candidates near the top of the list, so if the top 20 recommendations contain many genuinely valuable updates, the ranking is working well. In simple terms, a higher precision@20 means better prioritization.
